### Sequential Chains

```python
# Imports
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence
```

```python
# Load env
load_dotenv()
```

```python
# Model
model = ChatOpenAI(model='gpt-4.1-mini', max_completion_tokens=200)
```

```python
# Prompt Template to give the joke on topic
prompt = PromptTemplate(
    template="Give me only a joke on {topic}",
    input_variables={"topic"}
)
```

```python
# string Parser
parser = StrOutputParser()
```

```python
# Creating the chain using function

chain_using_function = RunnableSequence(prompt, model, parser)

# call the model
result = chain_using_function.invoke({
    'topic': 'AI'
})

print(result)
```

```python
# Creating the chain using LCEL

chain_using_lcel = prompt | model | parser

# call the model
result = chain_using_lcel.invoke({
    'topic': 'chatgpt vs gemini'
})

print(result)
```

#### Extending the Sequence

```python
# Now we will ask the model to explain the joke

extended_prompt = PromptTemplate(
    template="Explain the joke: {joke_txt}",
    input_variables={"joke_txt"}
)

extended_chain = chain_using_lcel | extended_prompt | model | parser

res = extended_chain.invoke({
    'topic': 'chatgpt vs gemini'
})

res
```

```python
extended_chain.get_graph().print_ascii()
```
