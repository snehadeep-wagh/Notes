```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
```

```python
directory_loader = DirectoryLoader(
    path='./documents',
    glob='*.pdf',
    loader_cls=PyPDFLoader
)
```

### lazy loader

```python
path = './documents'
```

```python
loader = DirectoryLoader(
    path=path,
    glob='*.pdf',
    loader_cls=PyPDFLoader
)
```

```python
lazy_res = loader.lazy_load()
```

```python
for res in lazy_res:
    print(res.page_content + "\n\n\n")
```
