### Creating the Tools
i) Using @tool decorator
ii) Using StructuredTools
iii) Using BaseTool class


Using tool decorator

```python
from langchain_core.tools import tool
```

```python
@tool
def multiply_with_decorator(a: int, b: int) -> int:
    """This function give the product of 2 variables."""
    return a * b
```

```python
ans = multiply_with_decorator.invoke({'a': '4', 'b': 5})
ans
```

```python
print(tool.name)
print(tool.description)
print(tool.args)
```

Using structured tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(description="The first number to multiply")
    b: int = Field(description="The second number to multiply")
```

```python
def multiply(a: int, b: int) -> int:
    return a * b
```

```python
tool = StructuredTool.from_function(
    func=multiply,
    description="This function give the product of 2 variables.",
    args_schema=MultiplyInput
)
```

```python
tool.invoke({'a':"5", 'b':4})
```

```python
print(tool.name)
print(tool.description)
print(tool.args)
```

Usinng BaseTool

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a: int = Field(description="The first number to add")
    b: int = Field(description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

Toolkit

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
